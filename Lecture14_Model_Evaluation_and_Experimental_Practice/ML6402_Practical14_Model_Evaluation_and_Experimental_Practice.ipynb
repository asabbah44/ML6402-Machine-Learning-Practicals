## Design a reproducible evaluation experiment

This practical moves beyond a single train/test score.

We use the breast-cancer dataset and compare models under a consistent evaluation protocol.

The key ideas are:

- keep preprocessing inside a pipeline;
- use stratified splits for classification;
- evaluate more than one metric;
- use cross-validation to estimate performance variability;
- compare models under the same folds whenever possible.

A single accuracy value can be misleading. Repeated or cross-validated evaluation gives a more reliable picture of model behavior.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

data=load_breast_cancer(as_frame=True)
X_train,X_test,y_train,y_test=train_test_split(
    data.data,data.target,test_size=0.2,random_state=42,stratify=data.target
)

models={
    "Logistic":make_pipeline(StandardScaler(),LogisticRegression(max_iter=2000)),
    "RandomForest":RandomForestClassifier(n_estimators=300,random_state=42)
}
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)

rows=[]
for name,model in models.items():
    scores=cross_validate(model,X_train,y_train,cv=cv,scoring=["accuracy","precision","recall","f1"])
    rows.append({"model":name,
                 "cv_accuracy":scores["test_accuracy"].mean(),
                 "cv_f1":scores["test_f1"].mean()})
pd.DataFrame(rows)
```
