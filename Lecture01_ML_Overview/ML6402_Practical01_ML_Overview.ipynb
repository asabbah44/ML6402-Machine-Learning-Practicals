- evaluate on unseen data

## Load a small machine-learning dataset

We use the built-in Iris dataset because it is simple, clean, and requires no external file.

The dataset contains numerical flower measurements and a categorical species target.

The first goal is not to build a sophisticated model. It is to understand the complete ML workflow:

1. load data;
2. separate features and target;
3. split into training and test sets;
4. fit a model;
5. make predictions;
6. evaluate the result.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

iris = load_iris(as_frame=True)
df = iris.frame
df.head()
```

## Build a first classification baseline

`X` contains the input features and `y` contains the target class.

The dataset is split into training and test sets so the model is evaluated on observations it did not use for fitting.

`LogisticRegression` is used as a simple baseline classifier.

The important distinction is:

- `fit()` learns model parameters from the training set;
- `predict()` applies the learned model to new observations;
- accuracy compares predicted classes with the true test labels.

This cell demonstrates the minimum end-to-end structure that will be reused throughout the course.

```python
X = df.drop(columns=["target"])
y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Train shape:", X_train.shape)
print("Test shape :", X_test.shape)
print("Accuracy   :", accuracy_score(y_test, y_pred))
```
