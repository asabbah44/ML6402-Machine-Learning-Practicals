## Case Study 1 – Load the Mall Customers dataset

The first source-chapter clustering example uses customer income and spending score.

Because clustering is unsupervised, there is no target variable. We are looking for natural groups in the feature space.

The dataset loader supports both local execution and GitHub/Colab fallback.

```python
from pathlib import Path
import pandas as pd

DATA_FILE = "Mall_Customers.csv"
LOCAL_CANDIDATES = [Path("data") / DATA_FILE, Path("Lecture09_Clustering") / "data" / DATA_FILE]
GITHUB_URL = "https://raw.githubusercontent.com/asabbah44/ML6402-Machine-Learning-Practicals/main/Lecture09_Clustering/data/Mall_Customers.csv"
local_path = next((p for p in LOCAL_CANDIDATES if p.exists()), None)
data_source = local_path if local_path else GITHUB_URL
df = pd.read_csv(data_source)
print(f"Loaded from: {data_source}")
df.head()
```

## Use the elbow method

K-Means requires the number of clusters `k` to be chosen in advance.

The elbow method fits several K-Means models and records the within-cluster sum of squares, exposed by Scikit-Learn as `inertia_`.

As `k` increases, inertia always decreases. We look for the point where the improvement begins to slow down—the approximate "elbow."

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X = df[["Annual Income (k$)", "Spending Score (1-100)"]]
inertias=[]
for k in range(1,11):
    km=KMeans(n_clusters=k,random_state=42,n_init=10)
    km.fit(X)
    inertias.append(km.inertia_)
plt.plot(range(1,11), inertias, marker="o")
plt.xlabel("k")
plt.ylabel("Inertia")
plt.show()
```

```python
kmeans=KMeans(n_clusters=5,random_state=42,n_init=10)
labels=kmeans.fit_predict(X)
plt.scatter(X.iloc[:,0],X.iloc[:,1],c=labels)
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],marker="X",s=200)
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.show()
```

## Case Study 1 – Load the Mall Customers dataset

The first source-chapter clustering example uses customer income and spending score.

Because clustering is unsupervised, there is no target variable. We are looking for natural groups in the feature space.

The dataset loader supports both local execution and GitHub/Colab fallback.

```python
from pathlib import Path
import pandas as pd

US_FILE = "USArrests.csv"
US_CANDIDATES = [
    Path("data") / US_FILE,
    Path("Lecture09_Clustering") / "data" / US_FILE
]
US_URL = "https://raw.githubusercontent.com/asabbah44/ML6402-Machine-Learning-Practicals/main/Lecture09_Clustering/data/USArrests.csv"

us_path = next((p for p in US_CANDIDATES if p.exists()), None)
us_source = us_path if us_path else US_URL
us_df = pd.read_csv(us_source)

print("Loaded from:", us_source)
us_df.head()
```

## Use the elbow method

K-Means requires the number of clusters `k` to be chosen in advance.

The elbow method fits several K-Means models and records the within-cluster sum of squares, exposed by Scikit-Learn as `inertia_`.

As `k` increases, inertia always decreases. We look for the point where the improvement begins to slow down—the approximate "elbow."

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X_us = us_df.iloc[:, 1:5].values

us_inertias = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, init="k-means++", random_state=0, n_init=10)
    km.fit(X_us)
    us_inertias.append(km.inertia_)

plt.plot(range(1, 11), us_inertias, marker="o")
plt.title("US Arrests – Elbow Method")
plt.xlabel("Number of clusters")
plt.ylabel("WCSS / Inertia")
plt.show()
```

```python
us_kmeans = KMeans(n_clusters=3, init="k-means++", random_state=0, n_init=10)
us_labels = us_kmeans.fit_predict(X_us)

plt.scatter(X_us[:, 0], X_us[:, 1], c=us_labels)
plt.scatter(
    us_kmeans.cluster_centers_[:, 0],
    us_kmeans.cluster_centers_[:, 1],
    marker="X",
    s=200
)
plt.title("US States – K-Means Clusters")
plt.xlabel("Murder")
plt.ylabel("Assault")
plt.show()
```

## Hierarchical clustering

The author chapter also uses a dendrogram with Ward linkage and then fits an agglomerative clustering model.

## Build a dendrogram

Hierarchical clustering does not require us to commit to a cluster count before constructing the hierarchy.

The dendrogram is generated using Ward linkage. It shows the sequence of cluster merges and the distance at which they occur.

Large vertical gaps can provide evidence for a reasonable number of clusters.

```python
import scipy.cluster.hierarchy as sch

plt.figure(figsize=(10, 4))
sch.dendrogram(sch.linkage(X_us, method="ward"))
plt.title("US Arrests – Dendrogram")
plt.xlabel("Data points")
plt.ylabel("Euclidean distance")
plt.show()
```

## Fit Agglomerative Clustering

After inspecting the dendrogram, the source example fits an agglomerative model.

`linkage="ward"` merges clusters in a way that minimizes the increase in within-cluster variance.

Unlike K-Means, hierarchical clustering does not learn centroids as part of the model; it builds groups through successive merges.

```python
from sklearn.cluster import AgglomerativeClustering

us_hc = AgglomerativeClustering(
    n_clusters=2,
    metric="euclidean",
    linkage="ward"
)
us_hc_labels = us_hc.fit_predict(X_us)

plt.scatter(X_us[:, 0], X_us[:, 1], c=us_hc_labels)
plt.title("US States – Agglomerative Clustering")
plt.xlabel("Murder")
plt.ylabel("Assault")
plt.show()
```

# Source Chapter Example 3 – Iris

The third author case study uses the Iris dataset and K-Means clustering.

## Case Study 3 – Iris clustering

The Iris case study uses the built-in Scikit-Learn dataset, so no external file is required.

We first calculate inertia for several values of `k`, then fit K-Means with three clusters because the source example is based on the three natural Iris groups.

The cluster labels are learned without using the true species labels.

```python
from sklearn.datasets import load_iris

iris = load_iris()
X_iris = iris.data

iris_inertias = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, init="k-means++", random_state=0, n_init=10)
    km.fit(X_iris)
    iris_inertias.append(km.inertia_)

plt.plot(range(1, 11), iris_inertias, marker="o")
plt.title("Iris – Elbow Method")
plt.xlabel("Number of clusters")
plt.ylabel("WCSS / Inertia")
plt.show()
```

```python
iris_kmeans = KMeans(n_clusters=3, init="k-means++", random_state=0, n_init=10)
iris_labels = iris_kmeans.fit_predict(X_iris)

plt.scatter(X_iris[:, 0], X_iris[:, 1], c=iris_labels)
plt.scatter(
    iris_kmeans.cluster_centers_[:, 0],
    iris_kmeans.cluster_centers_[:, 1],
    marker="X",
    s=200
)
plt.title("Iris – K-Means Clusters")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```
