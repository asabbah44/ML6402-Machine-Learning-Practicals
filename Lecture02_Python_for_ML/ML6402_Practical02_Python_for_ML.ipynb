## NumPy foundations for machine learning

NumPy arrays are the basic numerical structure used by many Python ML libraries.

This section introduces:

- creating arrays;
- checking shape;
- indexing and slicing;
- vectorized arithmetic;
- simple summary operations.

Vectorized operations are important because they avoid slow manual Python loops for numerical calculations.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

values = np.array([2, 4, 6, 8, 10])
print("Mean:", values.mean())
print("Std :", values.std())
```

## Create and inspect a Pandas DataFrame

Pandas is used for labeled tabular data.

This small student dataset demonstrates:

- creating a DataFrame;
- accessing columns;
- inspecting rows;
- working with mixed column types.

In real ML projects, most CSV datasets are first loaded into a Pandas DataFrame before preprocessing and modeling.

```python
students = pd.DataFrame({
    "study_hours": [2, 5, 3, 7, 4],
    "score": [55, 82, 65, 91, 74],
    "passed": [0, 1, 1, 1, 1]
})
students.head()
```

## Inspect data quality before modeling

`info()` summarizes:

- number of rows;
- column names;
- non-null counts;
- data types.

`describe()` summarizes numerical distributions.

`isna()` checks for missing values.

These are basic diagnostic steps that should be performed before transforming or modeling any dataset.

```python
students.info()
display(students.describe())
display(students.isna().sum())

plt.scatter(students["study_hours"], students["score"])
plt.xlabel("Study hours")
plt.ylabel("Score")
plt.show()
```
